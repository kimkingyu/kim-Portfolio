```python
import os
from moviepy.editor import VideoFileClip

# 获取当前工作目录（因为是在Jupyter Notebook中，使用os.getcwd() ）
current_dir = os.getcwd()

# 视频文件所在文件夹路径
cv_folder_path = os.path.join(current_dir, 'cv')

# 列出cv文件夹下所有视频文件
video_files = [f for f in os.listdir(cv_folder_path) if f.endswith('.mp4')]

# 构建视频文件的完整路径列表
video_paths = [os.path.join(cv_folder_path, file) for file in video_files]

# 获取视频时长
for path in video_paths:
    try:
        clip = VideoFileClip(path)
        duration = clip.duration
        print(f"{path} 的时长为: {duration} 秒")
        clip.close()
    except Exception as e:
        print(f"处理 {path} 时出现错误: {e}")
```
